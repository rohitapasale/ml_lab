```python
import numpy as np

# Hidden states
states = ["Rainy", "Sunny"]

# Possible observations
observations = ["Walk", "Shop", "Clean"]

# Initial probabilities
pi = np.array([0.6, 0.4])

# Transition matrix
A = np.array([
    [0.7, 0.3],
    [0.4, 0.6]
])

# Emission matrix
B = np.array([
    [0.1, 0.4, 0.5],
    [0.6, 0.3, 0.1]
])

# Observation sequence
# Walk = 0
# Shop = 1
# Clean = 2
obs = [0, 1, 2]

print("States:", states)
print("Observations:", observations)

print("\nInitial Probability:")
print(pi)

print("\nTransition Matrix:")
print(A)

print("\nEmission Matrix:")
print(B)

print("\nObservation Sequence:")
print(obs)
```

```python
def viterbi_algorithm(obs, A, B, pi):

    # Number of hidden states
    N = len(A)

    # Number of observations
    T = len(obs)

    # Viterbi probability table
    viterbi = np.zeros((T, N))

    # Backpointer table
    backpointer = np.zeros((T, N), dtype=int)

    # -------------------------
    # Step 1: Initialization
    # -------------------------

    viterbi[0] = pi * B[:, obs[0]]

    # -------------------------
    # Step 2: Recursion
    # -------------------------

    for t in range(1, T):

        for j in range(N):

            # Probability of coming to state j
            # from every previous state
            probabilities = (
                viterbi[t - 1] * A[:, j]
            )

            # Store the previous state
            # that gives maximum probability
            backpointer[t, j] = np.argmax(probabilities)

            # Store maximum probability
            # and multiply by emission probability
            viterbi[t, j] = (
                np.max(probabilities)
                * B[j, obs[t]]
            )

    # -------------------------
    # Step 3: Termination
    # -------------------------

    best_last_state = np.argmax(viterbi[T - 1])

    # -------------------------
    # Step 4: Backtracking
    # -------------------------

    best_path = [best_last_state]

    for t in range(T - 1, 0, -1):

        best_last_state = backpointer[t, best_last_state]

        best_path.append(best_last_state)

    # Reverse the path
    best_path.reverse()

    return viterbi, backpointer, best_path
```

```python
viterbi, backpointer, best_path = viterbi_algorithm(
    obs, A, B, pi
)

print("Viterbi Table:")
print(viterbi)

print("\nBackpointer Table:")
print(backpointer)

print("\nMost Likely Hidden State Sequence:")

for state in best_path:
    print(states[state], end=" -> ")
```
