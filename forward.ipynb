```python
import numpy as np

# Hidden states
states = ["Rainy", "Sunny"]

# Possible observations
observations = ["Walk", "Shop", "Clean"]

# Initial probabilities
pi = np.array([0.6, 0.4])

# Transition probability matrix
A = np.array([
    [0.7, 0.3],
    [0.4, 0.6]
])

# Emission probability matrix
B = np.array([
    [0.1, 0.4, 0.5],
    [0.6, 0.3, 0.1]
])

# Observation sequence
# Walk = 0, Shop = 1, Clean = 2
obs = [0, 1, 2]

print("States:", states)
print("Observations:", observations)
print("Initial Probability:", pi)
print("Transition Matrix:\n", A)
print("Emission Matrix:\n", B)
```

```python
def forward_algorithm(obs, A, B, pi):

    # Number of hidden states
    N = len(A)

    # Number of observations
    T = len(obs)

    # Create alpha table
    alpha = np.zeros((T, N))

    # Step 1: Initialization
    alpha[0] = pi * B[:, obs[0]]

    # Step 2: Recursion
    for t in range(1, T):

        for j in range(N):

            alpha[t, j] = (
                np.sum(alpha[t - 1] * A[:, j])
                * B[j, obs[t]]
            )

    # Step 3: Termination
    probability = np.sum(alpha[T - 1])

    return probability, alpha
```

```python
forward_probability, alpha = forward_algorithm(
    obs, A, B, pi
)

print("Forward Probability:", forward_probability)

print("\nAlpha Table:")
print(alpha)
```
